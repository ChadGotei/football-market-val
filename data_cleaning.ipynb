```python
# import libraries
# Import necessary libraries
import pandas as pd
import numpy as np
import unidecode
```

```python
# load all files
df_standard = pd.read_csv('FBref_datasets/top5_league_stats.csv')
df_shooting = pd.read_csv('FBref_datasets/top5_league_shooting.csv')
df_passing = pd.read_csv('FBref_datasets/top5_league_passing.csv')
df_pass_type = pd.read_csv('FBref_datasets/top5_league_passing_types.csv')
df_shot_creation = pd.read_csv('FBref_datasets/top5_league_gca.csv')
df_defense = pd.read_csv('FBref_datasets/top5_league_defense.csv')
df_possession = pd.read_csv('FBref_datasets/top5_league_possession.csv')
df_misc = pd.read_csv('FBref_datasets/top5_league_misc.csv')
df_playing_time = pd.read_csv('FBref_datasets/top5_league_playingtime.csv')

# df_standard
```

```python
# RK - s.no, matches -> matches 
def clean_data(df):
    df = df.drop(columns=['Rk', 'Matches'])
    return df
```

```python
# apply the clean data to each of data frame
df_standard = clean_data(df_standard)
df_shooting = clean_data(df_shooting)
df_passing = clean_data(df_passing)
df_pass_type = clean_data(df_pass_type)
df_shot_creation = clean_data(df_shot_creation)
df_defense = clean_data(df_defense)
df_possession = clean_data(df_possession)
df_misc = clean_data(df_misc)
df_playing_time = clean_data(df_playing_time)

df_standard
```

```python
# Function to remove duplicates of players who transfered mid season
def remove_duplicates(df):
    df = df.sort_values('Min', ascending=False).drop_duplicates(subset=['Player', 'Age'], keep='first').reset_index(drop=True)
    return df
```

```python
# combining all files into one
from functools import reduce

data_frames = [
    df_standard,
    df_shooting,
    df_passing,
    df_pass_type,
    df_shot_creation,
    df_defense,
    df_possession,
    df_misc,
    df_playing_time
]

merge_keys = [
    'Player', 'Nation', 'Pos', 'Squad',
    'Comp', 'Age', 'Born'
]

combined_df = data_frames[0].copy()

for df_temp in data_frames[1:]:
    duplicate_columns = [
        col for col in df_temp.columns
        if col in combined_df.columns and col not in merge_keys
    ]

    df_temp = df_temp.drop(columns=duplicate_columns)

    combined_df = pd.merge(
        combined_df,
        df_temp,
        on=merge_keys,
        how='left'
    )


combined_df = remove_duplicates(combined_df)
combined_df
```

```python
combined_df.rename(columns={"Squad": "Team"}, inplace=True)
```

```python
defensive_columns = [
    'Tkl%', 'TklW', 'Tkl', 'Tkl+Int',
    'Blocks', 'Int', 'Clr', 'Err', 'Won%'
]

# Keep only columns that exist in combined_df
defensive_columns = [
    col for col in defensive_columns
    if col in combined_df.columns
]

print("Defensive columns found:", defensive_columns)

for col in defensive_columns:
    missing_players = combined_df[combined_df[col].isna()]

    print(f"\n{col} — Missing values by position:")
    print(missing_players['Pos'].value_counts())
```

```python
#? dealing with nulls 
# tkl% and won% -> if forward then set to 0 if defender then take mean

def fill_defensive_nulls(df):
    df = df.copy()

    defensive_percent_cols = ['Tkl%', 'Won%']

    for col in defensive_percent_cols:
        if col in df.columns:
            # Mean among players whose position includes DF
            defender_mean = df.loc[
                df['Pos'].str.contains('DF', na=False) &
                df[col].notna(),
                col
            ].mean()

            # Non-defenders: missing defensive percentage = 0
            non_defender_mask = (
                ~df['Pos'].str.contains('DF', na=False) &
                df[col].isna()
            )
            df.loc[non_defender_mask, col] = 0

            # Defenders: missing value = defender-specific mean
            defender_mask = (
                df['Pos'].str.contains('DF', na=False) &
                df[col].isna()
            )
            df.loc[defender_mask, col] = round(defender_mean, 1)

    return df
```

```python
# to be replaced with mean as these are defenders stats and mostly defenders are missing it so to take better measure 
mean_columns = ['Succ%', 'Tkl%', 'Tkld%', 'Cmp%.3']

for col in mean_columns:
    combined_df[col] = combined_df[col].fillna(
        round(combined_df[col].mean(), 1)
    )
 
# Fill missing shot-related statistics with 0 -> since these belong to mostly defenders
zero_columns = ['SoT%', 'G/Sh', 'G/SoT', 'Dist', 'npxG/Sh', 'Mn/Start', 'Mn/Sub', 'On-Off', 'On-Off.1']

for col in zero_columns:
    combined_df[col] = combined_df[col].fillna(0)
    

# missing values after chages
combined_df[mean_columns + zero_columns].isnull().sum()
```

```python
# Load the transfer market dataset
tm_df = pd.read_csv('Transfermarkt _Datasets/top5_league_market_values.csv')
tm_df.drop(labels=tm_df.columns[0], axis=1, inplace=True)
tm_df.rename(columns={"Players": "Player"}, inplace=True)

# converting players name that contain letters such as ē,ā etc to ASCII equivalent
def clean_text(df, columns):
    for column in columns:
        df[column].fillna('', inplace=True)
        if column == 'Team':
             df[column] = df[column].apply(lambda x: unidecode.unidecode(x).lower())
        else:
           df[column] = df[column].apply(lambda x: unidecode.unidecode(x))
    return df

# Apply the cleaning function to both DataFrames
tm_df = clean_text(tm_df,['Player','Team'])
combined_df = clean_text(combined_df,['Player','Team'])

tm_df
```
